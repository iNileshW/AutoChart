```python
import re
import numpy as np
import pandas as pd
import geopandas as gpd

# 1) Align CRS and move to a metric CRS for area-based overlap scoring
metric_crs = "EPSG:27700"

if gdf_old.crs != gdf_new.crs:
    gdf_new_aligned = gdf_new.to_crs(gdf_old.crs)
else:
    gdf_new_aligned = gdf_new.copy()

old_m = gdf_old.to_crs(metric_crs).copy()
new_m = gdf_new_aligned.to_crs(metric_crs).copy()

old_m["old_row"] = old_m.index
new_m["new_row"] = new_m.index

# 2) Normalize names and scales for attribute checks
def norm_text(value):
    if pd.isna(value):
        return None
    text = str(value).strip().lower()
    text = re.sub(r"[^a-z0-9]+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text or None

old_m["old_name_norm"] = old_m.get("PANEL_MAIN", pd.Series(index=old_m.index)).apply(norm_text)
new_m["new_name_norm"] = new_m.get("Panel_Name", pd.Series(index=new_m.index)).apply(norm_text)

old_m["old_scale_num"] = pd.to_numeric(old_m.get("SCALE", np.nan), errors="coerce")
new_m["new_scale_num"] = pd.to_numeric(new_m.get("Pan_Scale", np.nan), errors="coerce")

# 3) Candidate pairs by spatial intersection
pairs = gpd.sjoin(
    old_m[["old_row", "PANEL_MAIN", "old_name_norm", "SCALE", "old_scale_num", "geometry"]],
    new_m[["new_row", "Panel_Name", "new_name_norm", "Pan_Scale", "new_scale_num", "geometry"]],
    how="inner",
    predicate="intersects",
)

if pairs.empty:
    match_table = pd.DataFrame(columns=[
        "old_row", "new_row", "PANEL_MAIN", "Panel_Name",
        "overlap_old_pct", "overlap_new_pct",
        "name_match_exact", "name_match_partial", "scale_match",
        "match_score", "confidence"
    ])
else:
    # 4) Compute intersection area for each candidate pair
    intersections = gpd.overlay(
        old_m[["old_row", "geometry"]],
        new_m[["new_row", "geometry"]],
        how="intersection",
    )
    intersections["intersect_m2"] = intersections.geometry.area

    overlap_area = (
        intersections.groupby(["old_row", "new_row"], as_index=False)["intersect_m2"]
        .sum()
    )

    old_area_map = old_m.set_index("old_row").geometry.area.to_dict()
    new_area_map = new_m.set_index("new_row").geometry.area.to_dict()

    match_table = (
        pairs.drop_duplicates(subset=["old_row", "new_row"])
        .merge(overlap_area, on=["old_row", "new_row"], how="left")
        .fillna({"intersect_m2": 0.0})
    )

    match_table["old_area_m2"] = match_table["old_row"].map(old_area_map)
    match_table["new_area_m2"] = match_table["new_row"].map(new_area_map)

    match_table["overlap_old_pct"] = (
        100 * match_table["intersect_m2"] / match_table["old_area_m2"].replace(0, np.nan)
    ).fillna(0).round(2)
    match_table["overlap_new_pct"] = (
        100 * match_table["intersect_m2"] / match_table["new_area_m2"].replace(0, np.nan)
    ).fillna(0).round(2)

    match_table["name_match_exact"] = (
        (match_table["old_name_norm"].notna())
        & (match_table["new_name_norm"].notna())
        & (match_table["old_name_norm"] == match_table["new_name_norm"])
    )

    match_table["name_match_partial"] = match_table.apply(
        lambda r: bool(r["old_name_norm"]) and bool(r["new_name_norm"]) and (
            r["old_name_norm"] in r["new_name_norm"] or r["new_name_norm"] in r["old_name_norm"]
        ),
        axis=1,
    )

    match_table["scale_match"] = np.isclose(
        match_table["old_scale_num"],
        match_table["new_scale_num"],
        atol=1,
        equal_nan=False,
    )

    # Weighted score: geometry-first, then name and scale agreement
    match_table["match_score"] = (
        0.55 * (match_table["overlap_old_pct"] / 100.0)
        + 0.25 * (match_table["overlap_new_pct"] / 100.0)
        + 0.15 * match_table["name_match_exact"].astype(float)
        + 0.03 * match_table["name_match_partial"].astype(float)
        + 0.02 * match_table["scale_match"].astype(float)
    ).round(4)

    match_table["confidence"] = pd.cut(
        match_table["match_score"],
        bins=[-np.inf, 0.25, 0.50, 0.75, np.inf],
        labels=["low", "medium", "high", "very_high"],
    )

    match_table = match_table.sort_values(
        by=["match_score", "overlap_old_pct", "overlap_new_pct"],
        ascending=False,
    )

# Final robust comparison table
match_table = match_table[[
    "old_row", "new_row", "PANEL_MAIN", "Panel_Name", "SCALE", "Pan_Scale",
    "overlap_old_pct", "overlap_new_pct",
    "name_match_exact", "name_match_partial", "scale_match",
    "match_score", "confidence",
]].reset_index(drop=True)

match_table.head(25)
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
from matplotlib.lines import Line2D

def plot_panel_overlap(panel_main, old_df=None, new_df=None, metric_crs="EPSG:27700", figsize=(9, 9)):
    """
    Plot old panel geometry, intersecting new geometries, and overlap area for a PANEL_MAIN value.
    Returns: (ax, metrics_dict)
    """
    if old_df is None:
        old_df = gdf_old
    if new_df is None:
        new_df = gdf_new

    if "PANEL_MAIN" not in old_df.columns:
        raise KeyError("old_df must contain a 'PANEL_MAIN' column.")
    if "geometry" not in old_df.columns or "geometry" not in new_df.columns:
        raise KeyError("Both dataframes must contain a geometry column.")

    panel_value = str(panel_main).strip()
    old_mask = old_df["PANEL_MAIN"].astype(str).str.strip().str.casefold() == panel_value.casefold()
    old_selected = old_df.loc[old_mask].copy()

    if old_selected.empty:
        raise ValueError(f"No old polygons found for PANEL_MAIN='{panel_main}'.")

    # Align CRS before metric reprojection
    if old_df.crs != new_df.crs:
        new_df = new_df.to_crs(old_df.crs)

    old_target = gpd.GeoDataFrame(
        {"PANEL_MAIN": [panel_value]},
        geometry=[old_selected.geometry.unary_union],
        crs=old_df.crs,
    )

    old_m = old_target.to_crs(metric_crs)
    new_m = new_df.to_crs(metric_crs).copy()

    # Candidate new polygons are those spatially intersecting the old target geometry
    new_hits = new_m[new_m.intersects(old_m.geometry.iloc[0])].copy()

    if new_hits.empty:
        fig, ax = plt.subplots(figsize=figsize)
        old_m.boundary.plot(ax=ax, color="navy", linewidth=2)
        ax.set_title(f"{panel_value}: no intersecting new polygons found")
        ax.set_axis_off()
        plt.show()
        return ax, {"old_area_m2": float(old_m.geometry.area.iloc[0]), "new_overlap_area_m2": 0.0, "overlap_pct_old": 0.0}

    overlap = gpd.overlay(
        old_m[["PANEL_MAIN", "geometry"]],
        new_hits[["geometry"]],
        how="intersection",
    )

    old_area = float(old_m.geometry.area.iloc[0])
    overlap_area = float(overlap.geometry.area.sum()) if not overlap.empty else 0.0
    overlap_pct_old = (100.0 * overlap_area / old_area) if old_area > 0 else 0.0

    fig, ax = plt.subplots(figsize=figsize)
    old_m.boundary.plot(ax=ax, color="navy", linewidth=2)
    new_hits.boundary.plot(ax=ax, color="darkorange", linewidth=1.2, alpha=0.9)

    if not overlap.empty:
        overlap.plot(ax=ax, color="limegreen", alpha=0.55, edgecolor="none")

    legend_items = [
        Line2D([0], [0], color="navy", lw=2, label="Old panel boundary"),
        Line2D([0], [0], color="darkorange", lw=1.2, label="Intersecting new boundaries"),
        Patch(facecolor="limegreen", alpha=0.55, label="Overlap area"),
    ]
    ax.legend(handles=legend_items, loc="best")
    ax.set_title(
        f"Overlap comparison for PANEL_MAIN='{panel_value}'\n"
        f"Overlap area: {overlap_area:,.0f} m^2 ({overlap_pct_old:.2f}% of old area)"
    )
    ax.set_axis_off()
    plt.tight_layout()
    plt.show()

    metrics = {
        "old_area_m2": old_area,
        "new_overlap_area_m2": overlap_area,
        "overlap_pct_old": overlap_pct_old,
        "new_polygons_intersecting": int(len(new_hits)),
    }
    return ax, metrics

# Example usage:
# ax, stats = plot_panel_overlap("Isles of Scilly Northern Part")
# stats
```

```python
import numpy as np
import pandas as pd
import geopandas as gpd


def match_chart_panels(chart_id: str, max_scale: int = 30000) -> pd.DataFrame:
    """
    Compare old and new chart panels for a given chart_id.
    New-panel candidates are limited to Pan_Scale <= max_scale (default 30000).
    Returns a DataFrame with overlap percentage and scale match per old panel.
    """
    if "gdf_old" not in globals():
        globals()["gdf_old"] = gpd.read_file(r"data_original/Old_Chart_Scheme.shp")
    if "gdf_new" not in globals():
        globals()["gdf_new"] = gpd.read_file(r"data_original/New_Chart_Scheme.shp")

    gdf_old_local = globals()["gdf_old"]
    gdf_new_local = globals()["gdf_new"]

    metric_crs = "EPSG:27700"

    old_panels = gdf_old_local[
        gdf_old_local["PANEL_IDEN"].astype(str).str.contains(chart_id, na=False)
    ].copy()
    if old_panels.empty:
        raise ValueError(f"No old panels found for chart {chart_id} in PANEL_IDEN.")

    new_panels = gdf_new_local[
        pd.to_numeric(gdf_new_local["Pan_Scale"], errors="coerce") <= max_scale
    ].copy()

    if gdf_old_local.crs != new_panels.crs:
        new_panels = new_panels.to_crs(gdf_old_local.crs)

    old_m = old_panels.to_crs(metric_crs).copy()
    new_m = new_panels.to_crs(metric_crs).copy()

    old_m["old_row"] = old_m.index
    new_m["new_row"] = new_m.index

    empty_cols = [
        "PANEL_IDEN", "PANEL_MAIN", "old_scale",
        "new_chart", "new_panel", "new_scale",
        "overlap_old_pct", "scale_match",
    ]

    if new_m.empty:
        return pd.DataFrame(columns=empty_cols)

    pairs = gpd.sjoin(
        old_m[["old_row", "PANEL_IDEN", "PANEL_MAIN", "SCALE", "geometry"]],
        new_m[["new_row", "Chart", "Panel_Name", "Pan_Scale", "geometry"]],
        how="inner",
        predicate="intersects",
    )

    if pairs.empty:
        return pd.DataFrame(columns=empty_cols)

    intersections = gpd.overlay(
        old_m[["old_row", "geometry"]],
        new_m[["new_row", "geometry"]],
        how="intersection",
    )
    intersections["intersect_m2"] = intersections.geometry.area

    overlap = (
        intersections.groupby(["old_row", "new_row"], as_index=False)["intersect_m2"]
        .sum()
    )

    old_area = old_m.set_index("old_row").geometry.area.rename("old_area_m2")

    compared = (
        pairs.drop_duplicates(subset=["old_row", "new_row"])
        .merge(overlap, on=["old_row", "new_row"], how="left")
        .merge(old_area, on="old_row", how="left")
        .fillna({"intersect_m2": 0.0})
    )

    compared["overlap_old_pct"] = (
        100 * compared["intersect_m2"] / compared["old_area_m2"].replace(0, np.nan)
    ).fillna(0).round(2)

    compared["old_scale_num"] = pd.to_numeric(compared["SCALE"], errors="coerce")
    compared["new_scale_num"] = pd.to_numeric(compared["Pan_Scale"], errors="coerce")
    compared["scale_match"] = np.isclose(
        compared["old_scale_num"], compared["new_scale_num"], atol=1, equal_nan=False
    )

    best = compared.sort_values(
        by=["old_row", "overlap_old_pct", "scale_match"],
        ascending=[True, False, False],
    ).drop_duplicates(subset=["old_row"], keep="first")

    return best[[
        "PANEL_IDEN", "PANEL_MAIN", "SCALE",
        "Chart", "Panel_Name", "Pan_Scale",
        "overlap_old_pct", "scale_match",
    ]].rename(columns={
        "SCALE": "old_scale",
        "Chart": "new_chart",
        "Panel_Name": "new_panel",
        "Pan_Scale": "new_scale",
    }).reset_index(drop=True)



```
